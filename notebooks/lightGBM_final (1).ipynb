```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
dataset = pd.read_csv("/content/reddit_preprocessing.csv")

# drop row with nwn values in "clean_comment"
cleaned_dataset = dataset.dropna()
```

```python
# serparate features and target
X_cleaned = cleaned_dataset['clean_comment']
y_cleaned = cleaned_dataset['category']
```

```python
## Split the cleaned data into train and test sets(80-20 split)
X_train_cleaned , X_test_cleaned , y_train_cleaned , y_test_cleaned = train_test_split(X_cleaned , y_cleaned, test_size=0.2,random_state=42)
```

```python
# Apply Tfidfvectoroiizer with trigram setting and max_features=1000
tfidf_cleaned = TfidfVectorizer(ngram_range=(1,3),max_features=1000)
```

```python
## fit the vectorizer on the training data and transform both train and test set
X_train_tfidf_cleaned = tfidf_cleaned.fit_transform(X_train_cleaned)
X_test_tfidf_cleaned = tfidf_cleaned.transform(X_test_cleaned)
```

```python
import lightgbm as lgb
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import classification_report, accuracy_score
from sklearn.model_selection import GridSearchCV
import optuna
```

```python
## FUnction to optimize LightGBM hyperparameter
def objective(trial):
  #Define hperparameter to be tuned
  param = {
      'objective': 'multiclass',
      'num_class': 3, # assuming 3 categories (-1 ,0,1)
      'learning_rate': trial.suggest_float("learning_rate", 1e-3 , 1e-1),
      'n_estimators': trial.suggest_int("n_estimators", 50, 500),
      'max_depth': trial.suggest_int("max_depth", 3, 20),
      'metric' : 'multi_logloss',
      'is_unbalance': True,
      'class_weight' : 'balanced'
    }

  # Define the lightGBM with the trial parmeters
  model = lgb.LGBMClassifier(**param)

  ## Perform cross-valiadation
  # Note: `cross_val_score` is from `sklearn.model_selection`
  # and needs to be imported or ensured it's in scope.
  scores = cross_val_score(model , X_train_tfidf_cleaned, y_train_cleaned, cv=3 , scoring = 'accuracy')

  # return the average score folds
  return scores.mean()
```

```python
# create an optuna study to optimize the hpyerparmater
study  = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=50)
```

```python
# extract the best hyperprameter
best_params = study.best_params
best_params
```

```python
best_model = lgb.LGBMClassifier(
    objective='multiclass',
    num_class=3,
    metric='multi_logloss',
    is_unbalance=True,
    class_weight='balanced',
    reg_alpha  = 0.1 , ## l1 regularization
    reg_lambda = 0.1 , ## l2 regularization
    leaning_rate  = 0.08,
    max_depth = 20,
    n_estimators = 367

)
```

```python
# fit the model on the resampled training data
best_model.fit(X_train_tfidf_cleaned,y_train_cleaned)
```

```python
# Predict on the train set
y_train_pred = best_model.predict(X_train_tfidf_cleaned)
```

```python
## calculate the accuracy on the test set
accuracy_train = accuracy_score(y_train_cleaned, y_train_pred)
print(f"Accuracy on train set: {accuracy_train}")
```

```python
## Generate classification report
report_train = classification_report(y_train_cleaned , y_train_pred)
print(report_train)
```

```python
## Predict on the test set
y_pred = best_model.predict(X_test_tfidf_cleaned)
```

```python
# calculate accuracy on the test set
accuracy_test = accuracy_score(y_test_cleaned, y_pred)
print(f"Accuracy on test set: {accuracy_test}")
```

```python
## generate classification report
report = classification_report(y_test_cleaned, y_pred)
print(report)
```

```python
import re
import numpy as np


# assuming you have pre-trained tfidf_vectorizer anda lgb_model loaded
# tfidf_vectorizer : your trained TF-IDF vectorizer
# lgbm_model : your tained lightGBM model

# Function to clean and preprocess a Youtube comment
def preprocess_comment(comment):
  #lowercasing
  comment = comment.lower()

  # Remove specical characters, urls , punctuation and extra space
  comment = re.sub(r"http\S+|www\S+|https\S+" , '' , comment , flags =re.MULTILINE)  # remove urls
  comment = re.sub(r'\W' , ' ' ,comment)  ## remove special characters
  comment = re.sub(r'\s+', ' ' ,comment).strip()  # remove extra space and newlines
  return comment


# prediction function
def predict_sentiment(comment , tfidf_veectorizer , lgbm_model):
  # step 1: Preprocess the YouTube comment
  cleaned_comment= preprocess_comment(comment)

  # step 2: transform the comment using the trained tF-idf vectorizer
  comment_tfidf = tfidf_veectorizer.transform([cleaned_comment])

  # step 3: use the trianed LighGBM model to predict the sentiment
  prediction  = lgbm_model.predict(comment_tfidf)
  prediction_proba = lgbm_model.predict_proba(comment_tfidf)

  # Step 4: Get the predicted sentiment (label) and probability
  sentoment_class = np.argmax(prediction_proba)
  sentiment_proba = np.max(prediction_proba)

  # step 5: Return the sentiment label and confidence
  return {
      'sentiment_class' : int(prediction[0])   ,  #-1 , 0  , or 1 depending on your labels
      'confidence' : sentiment_proba
  }

# example usage :
comment = "I absolutely love this video!"
result  = predict_sentiment(comment , tfidf_cleaned , best_model)
print(f"Predicted sentiment: {result['sentiment_class']},Confidence:{result['confidence']}")

```
